# Day 22 — Dataset Understanding
**Project 1:** Retail Sales & Customer Performance Analysis  
**Institute task:** inspect rows, columns, dtypes, unique values, missingness, duplicates, and create a data dictionary.  
**Source:** Original 30-row retail sales CSV. No cleaning is performed here.

```python
import pandas as pd
import numpy as np
from pathlib import Path
pd.set_option("display.max_columns", 30)
DATA_PATH = Path("retail_sales.csv")
df = pd.read_csv(DATA_PATH)
print("Loaded:", DATA_PATH, "| Shape:", df.shape)
```

## 1. Inspect the dataset — head, tail, shape, columns

```python
display(df.head())
display(df.tail())
print("Shape:", df.shape)
print("Columns:", df.columns.tolist())
```

## 2. Data types and structure — info, dtypes, describe

```python
df.info()
print("\nData types:")
display(df.dtypes.to_frame("dtype"))
print("\nNumerical summary:")
display(df.describe())
print("\nCategorical summary:")
display(df.describe(include="object"))
```

## 3. Analyze each column — unique values, missing values, examples

```python
meanings = {'Order_ID': 'Unique identifier of an order', 'Order_Date': 'Date on which the order was placed', 'Customer_Name': 'Customer display name; not a verified unique customer ID', 'City': 'City of the transaction (Region not supplied)', 'Category': 'Product category', 'Product': 'Product name', 'Quantity': 'Units purchased', 'Unit_Price': 'Price per unit in INR', 'Sales': 'Recorded sales revenue in INR', 'Customer_Rating': 'Customer rating; missing for some orders'}
profile = pd.DataFrame([{ "Column":c, "Meaning":meanings[c], "Data_Type":str(df[c].dtype), "Unique_Values":int(df[c].nunique(dropna=True)), "Missing_Count":int(df[c].isna().sum()), "Missing_Percent":round(100*df[c].isna().mean(),2), "Examples":", ".join(map(str,df[c].dropna().unique()[:3])) } for c in df.columns])
display(profile)
```

## 4. Unique categorical values and frequencies
The institute examples refer to `Region`; the actual CSV has `City`, so City is used here.

```python
for column in ["Category", "City", "Customer_Name", "Product"]:
    print(f"\n{column} unique values:", df[column].unique())
    display(df[column].value_counts(dropna=False).rename("Count").to_frame())
```

## 5. Missing values — count and percentage

```python
missing = pd.DataFrame({"Missing_Count":df.isnull().sum(),"Missing_Percent":(df.isnull().sum()/len(df)*100).round(2)})
display(missing)
print("Total missing cells:", int(df.isnull().sum().sum()))
```

## 6. Duplicate records

```python
print("Duplicate full rows:", int(df.duplicated().sum()))
print("Duplicate Order_ID values:", int(df["Order_ID"].duplicated().sum()))
display(df[df.duplicated(keep=False)])
```

## 7. Data availability and integrity checks
These are **observations only**. Data cleaning is reserved for Day 23.

```python
print("Order_Date raw dtype:", df["Order_Date"].dtype)
print("Sales total (INR):", int(df["Sales"].sum()))
print("Quantity total:", int(df["Quantity"].sum()))
print("Sales matches Quantity × Unit_Price:", bool((df["Sales"] == df["Quantity"]*df["Unit_Price"]).all()))
for col in ["Region","Profit","Customer_ID","Product_ID"]:
    print(f"{col} present?", col in df.columns)
```
