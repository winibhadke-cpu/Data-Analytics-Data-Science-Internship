# Evidence and recommendations

```python
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv('transformed_retail_sales.csv')
print(df.shape, df.Sales.sum())
for col in ['Product','City','Customer_Name','Category']:
    print(df.groupby(col).Sales.agg(['sum','count']))
print(df.groupby('Product').Quantity.sum())
print(df.Sales.describe())
print(df.groupby('Customer_Name').Order_ID.nunique())
print(df.groupby('Customer_Segment').Sales.sum())
print(df.Month_Name.unique(),df.Profit_Margin.notna().sum())
assert df.Sales.sum()==81870
for col in ['Product','City']:
    df.groupby(col).Sales.sum().sort_values().plot.barh(title=col+' revenue',color='#008C95')
    plt.xlabel('Revenue (INR)'); plt.tight_layout(); plt.savefig(col+'_revenue.png'); plt.show(); plt.close()
display(pd.read_csv('Business_Recommendations.csv'))
```
