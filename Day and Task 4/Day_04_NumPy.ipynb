## 1. Import NumPy and create a one-dimensional sales array
NumPy arrays store homogeneous numeric data efficiently. `ndim` gives the number of dimensions, `shape` the length along each axis, and `size` the number of elements.

```python
import numpy as np

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
sales = np.array([12000, 15000, 13500, 18000, 22000, 19500, 24000, 21000, 26000, 23000, 28000, 30000], dtype=float)
print("Sales (INR):", sales)
print("Dimensions:", sales.ndim, "| Shape:", sales.shape, "| Size:", sales.size, "| Dtype:", sales.dtype)
```

## 2. Indexing and slicing
Indices start at zero. `sales[0]` accesses January; `sales[-1]` accesses December. A slice such as `sales[:3]` returns January–March.

```python
print("January:", sales[0])
print("December:", sales[-1])
print("First quarter:", sales[:3])
print("Last quarter:", sales[-3:])
```

## 3. Required sales statistics
NumPy uses **population standard deviation** by default (`ddof=0`); this report uses that convention. The 25th and 75th percentiles use NumPy’s default linear interpolation.

```python
metrics = {
    "Total Sales": np.sum(sales),
    "Average Sales": np.mean(sales),
    "Maximum Sales": np.max(sales),
    "Minimum Sales": np.min(sales),
    "Standard Deviation (population)": np.std(sales),
    "Median": np.median(sales),
    "25th Percentile": np.percentile(sales, 25),
    "75th Percentile": np.percentile(sales, 75),
}
for name, value in metrics.items():
    print(f"{name}: INR {value:,.2f}")
```

## 4. Arithmetic operations and aggregations
Element-wise arithmetic can model an increase in monthly sales. Compare original totals with a hypothetical 10% growth scenario.

```python
sales_growth = sales * 1.10
print("10% growth scenario:", sales_growth)
print("Total projected sales:", round(np.sum(sales_growth), 2))
print("Monthly difference:", sales_growth - sales)
```

## 5. Two-dimensional arrays and reshaping
Reshape 12 monthly values into four quarters of three months each. The underlying data order is preserved. Aggregate along `axis=1` to get quarterly totals.

```python
quarterly = sales.reshape(4, 3)
print("2D quarterly array:\n", quarterly)
print("Dimensions:", quarterly.ndim, "Shape:", quarterly.shape)
print("Quarterly totals:", np.sum(quarterly, axis=1))
print("Quarter 2, month 1:", quarterly[1, 0])
print("First two quarters:\n", quarterly[:2, :])
```

## 6. Matrix multiplication
Matrix multiplication uses `@` (or `np.matmul`), **not** `*`, which multiplies element by element. Here, quantities for two products across three periods are multiplied by product prices to calculate period revenues.

```python
quantities = np.array([[10, 5], [8, 7], [12, 6]])  # 3 periods x 2 products
prices = np.array([[500], [800]])  # 2 products x 1 price vector
revenue = quantities @ prices
print("Quantities:\n", quantities)
print("Prices:\n", prices)
print("Period revenue (INR):\n", revenue)
print("Total revenue:", np.sum(revenue))
```

## 7. Interpretation and conclusions
- **Total** indicates cumulative sales over all 12 months.
- **Mean and median** summarize a typical month.
- **Maximum and minimum** show the range of monthly performance.
- **Standard deviation** describes how widely months vary around the mean.
- **Percentiles** describe the lower and upper portions of the distribution.
- **Quarterly totals** make period-to-period comparisons easier.
- **Matrix multiplication** combines product quantities and prices into revenue.
