## 1. Import Pandas and load the CSV
`pd.read_csv()` reads a comma-separated file into a DataFrame. `display()` shows a formatted table.

```python
import pandas as pd
from pathlib import Path

csv_path = Path("Day05_Retail_Sales_Dataset.csv")
if not csv_path.exists():
    csv_path = Path("/mnt/data/Day05_Retail_Sales_Dataset.csv")
df = pd.read_csv(csv_path)
print("Dataset loaded successfully")
print("File:", csv_path.name)
display(df.head())
```

## 2. Pandas Series and DataFrame
A **Series** is one labeled column. A **DataFrame** is a two-dimensional table of rows and columns.

```python
sales_series = df["Sales"]
print("Series type:", type(sales_series).__name__)
print("DataFrame type:", type(df).__name__)
print("First five sales values:")
print(sales_series.head().to_string(index=False))
```

## 3. Display the first 10 rows
`head(10)` is useful for quickly checking column values and structure.

```python
display(df.head(10))
```

## 4. Display the last 10 rows
`tail(10)` checks the final records.

```python
display(df.tail(10))
```

## 5. Number of rows and columns
`shape` returns a tuple: `(rows, columns)`.

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("Shape:", df.shape)
```

## 6. Column names and data types
`columns` lists the fields; `dtypes` gives their stored types. `info()` also shows non-null counts and memory usage.

```python
print("Columns:", list(df.columns))
print("\nData types:")
print(df.dtypes.to_string())
print("\nDataFrame information:")
df.info()
```

## 7. Identify missing values
`isnull().sum()` counts blanks in each column. Here, three ratings are deliberately blank; we only identify them, since cleaning belongs to Day 6.

```python
missing = df.isnull().sum()
print("Missing values by column:")
print(missing.to_string())
print("Total missing values:", int(missing.sum()))
print("Rows with missing ratings:")
display(df[df["Customer_Rating"].isnull()][["Order_ID", "Customer_Rating"]])
```

## 8. Generate summary statistics
`describe()` summarizes numerical columns using count, mean, standard deviation, min, quartiles and max. `describe(include="all")` also includes text fields.

```python
print("Numerical summary:")
display(df.describe().round(2))
print("All-column summary:")
display(df.describe(include="all").fillna("—"))
```

## 9. Additional beginner-friendly analysis
These operations are optional enrichment, showing why Pandas helps analysts understand business performance.

```python
print("Total sales:", int(df["Sales"].sum()))
print("Average order sales:", round(df["Sales"].mean(), 2))
print("Sales by category:")
print(df.groupby("Category")["Sales"].sum().sort_values(ascending=False).to_string())
```
