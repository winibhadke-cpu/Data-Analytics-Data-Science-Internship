```python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display
root=Path('../Customer_Churn_Prediction')
raw=root/'data/raw/customer_churn.csv'
assert raw.exists(), 'Add original customer_churn.csv to Customer_Churn_Prediction/data/raw first'
df=pd.read_csv(raw,dtype={'CustomerID':'string'})
display(df.head()); print(df.shape,df.columns.tolist()); df.info(); display(df.dtypes,df.describe(include='all'),df.isna().sum())
print('Numerical:',df.select_dtypes('number').columns.tolist()); print('Categorical:',df.select_dtypes(exclude='number').columns.tolist())
print('Duplicate IDs:',df.CustomerID.duplicated().sum()); print('Missing IDs:',df.CustomerID.isna().sum())
counts=df.Churn.value_counts(dropna=False); display(counts,counts/len(df)*100)
counts.plot.bar(title='Target distribution'); plt.ylabel('Customers'); plt.tight_layout(); plt.savefig('Target_Distribution.png'); plt.show()
meanings={'CustomerID':'Customer identifier','Churn':'Whether customer left','TenureMonths':'Months of service','ContractType':'Contract duration/type','MonthlyCharges':'Monthly payment','PaymentMethod':'Payment method','AvgMonthlyDataUsageGB':'Average monthly data usage in GB','SupportCallsLast90Days':'Support calls in last 90 days'}
dictionary=pd.DataFrame({'Column':df.columns,'Meaning':[meanings.get(c,'Confirm with source documentation: '+c) for c in df.columns],'Data Type':df.dtypes.astype(str).values,'Role':['Identifier' if c=='CustomerID' else 'Target' if c=='Churn' else 'Candidate feature' for c in df.columns]})
dictionary.to_excel('Data_Dictionary.xlsx',index=False); display(dictionary)
```
